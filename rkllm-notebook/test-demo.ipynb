```c++
#include <cstddef>

using std::size_t;

#include "/home/milh/projects/rknn-llm/rkllm-runtime/Linux/librkllm_api/include/rkllm.h"
#include <iostream>
#include <string>

static int callback(
    RKLLMResult* result,
    void* userdata,
    LLMCallState state)
{
    if (result && result->text)
        std::cout << result->text << std::flush;

    if (state == RKLLM_RUN_FINISH)
        std::cout << std::endl;

    return 0;
}

LLMHandle llm = nullptr;
```

```c++
RKLLMParam param = rkllm_createDefaultParam();

param.model_path =
"/home/milh/models/Qwen3-4B-w8a8-npu.rkllm";

int ret = rkllm_init(
    &llm,
    &param,
    callback);

std::cout << "init=" << ret << std::endl;
```

```c++
std::string prompt =
"Write a haiku about RK3588.";

RKLLMInput input{};
input.input_type = RKLLM_INPUT_PROMPT;
input.prompt_input = prompt.data();

RKLLMInferParam infer{};
infer.mode = RKLLM_INFER_GENERATE;
infer.keep_history = 1;

rkllm_run(llm, &input, &infer, nullptr);
```

```c++
auto start = std::chrono::steady_clock::now();

rkllm_run(llm, &input, &infer, nullptr);

auto end = std::chrono::steady_clock::now();

std::cout
    << std::chrono::duration_cast<std::chrono::milliseconds>(
           end - start)
           .count()
    << " ms\n";
```

```c++
rkllm_destroy(llm);
```
