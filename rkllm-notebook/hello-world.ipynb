```c++
#include "/home/milh/projects/rknn-llm/rkllm-runtime/Linux/librkllm_api/include/rkllm.h"
#include <cstring>
#include <iostream>

static int callback(
    RKLLMResult* result,
    void* userdata,
    LLMCallState state)
{
    if (result && result->text)
        std::cout << result->text << std::flush;

    if (state == RKLLM_RUN_FINISH)
        std::cout << "\n[FINISHED]\n";

    if (state == RKLLM_RUN_ERROR)
        std::cout << "\n[ERROR]\n";

    return 0;
}

LLMHandle handle = nullptr;

// Replace with your .rkllm model
const char* model_path =
    "/home/milh/models/Qwen3-4B-w8a8-npu.rkllm";

RKLLMParam param = rkllm_createDefaultParam();

param.model_path = model_path;
param.max_context_len = 4096;
param.max_new_tokens = 128;
param.top_k = 1;
param.top_p = 0.9f;
param.temperature = 0.8f;

int ret = rkllm_init(
    &handle,
    &param,
    callback);

std::cout << "rkllm_init = "
          << ret << std::endl;

if (ret != 0)
    return;

RKLLMInput input{};
input.input_type = RKLLM_INPUT_PROMPT;
input.prompt_input = (char*)"Say hello world.";

RKLLMInferParam infer{};
infer.mode = RKLLM_INFER_GENERATE;
infer.keep_history = 0;

ret = rkllm_run(
    handle,
    &input,
    &infer,
    nullptr);

std::cout << "\nrkllm_run = "
          << ret << std::endl;

rkllm_destroy(handle);
```
